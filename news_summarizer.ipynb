```python
import requests
import warnings
```

```python
warnings.filterwarnings('ignore')
```

```python
key='46f36cbf75014020b86bd80cabb8ae64'
url=f'https://newsapi.org/v2/everything?q=apple&from=2025-04-16&to=2025-04-16&sortBy=popularity&apiKey={key}'
```

```python
response=requests.get(url)
```

```python
articles = response.json().get("articles", [])
```

```python
if not articles:
    print("No news found.")
    exit()
```

```python
title = articles[0]['title']
content = articles[0]['content'] or articles[0]['description'] or "No content available."
text = f"{title}. {content}"
```

```python
print("📰 News Article:\n", text)
```

```python
from transformers import BartForConditionalGeneration, BartTokenizer
```

```python
model = BartForConditionalGeneration.from_pretrained("bart_model")
tokenizer = BartTokenizer.from_pretrained("bart_model")
```

```python
inputs = tokenizer([text], max_length=1024, return_tensors="pt", truncation=True)
summary_ids = model.generate(
    inputs["input_ids"],
    max_length=100,
    min_length=30,
    length_penalty=2.0,
    num_beams=4,
    early_stopping=True
)
```

```python
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
```

```python
print("Summary:\n", summary)
```

```python
import pyttsx3
```

```python
engine=pyttsx3.init()
```

```python
engine.say(summary)
engine.runAndWait()
```
